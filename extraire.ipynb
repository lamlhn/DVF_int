```python
import pandas as pd
```

```python
csv = pd.read_csv('./../dvf.csv')
```

```python
data = csv.dropna(subset=["valeur_fonciere"])
```

```python
df_unique = data[['code_departement', 'id_mutation']].drop_duplicates()

# 2. Gom nhóm theo departement và lấy tối đa 100 id_mutation đầu tiên của mỗi nhóm
kq_dataframe = df_unique.groupby('code_departement').head(500)

# Xem thử kết quả dạng DataFrame
print(kq_dataframe)
```

```python
len(kq_dataframe)
```

```python
data_filtered = data[data['id_mutation'].isin(kq_dataframe['id_mutation'])]
```
